```python
# Importation et Exploration du jeu de données
Ce notebook a pour but de charger le dataset Breast Cancer Wisconsin et de vérifier sa structure.
Je commence par inspecter les dimensions et m’assurer de l’absence de valeurs manquantes.
```

```python
# J'importe le jeu de données et je le mets au format DataFrame pour explorer facilement.
from sklearn.datasets import load_breast_cancer
import pandas as pd
import numpy as np

data = load_breast_cancer()
X = data.data
y = data.target
feature_names = data.feature_names

df = pd.DataFrame(X, columns=feature_names)
df['target'] = y

# Je regarde les premières lignes pour vérifier la distribution des variables
print('Premières lignes :')
print(df.head())

# Je vérifie la répartition des classes : c’est important pour l’équilibre du modèle
print('Classes cibles :', data.target_names)
print('Distribution des classes :', np.bincount(y))

# Je teste l’existence de valeurs manquantes, ce serait bloquant pour la suite
print('Valeurs manquantes dans le dataset :', df.isnull().sum().sum())
```

J’observe que les données sont propres, sans valeurs manquantes, et la classification binaire est bien équilibrée.
