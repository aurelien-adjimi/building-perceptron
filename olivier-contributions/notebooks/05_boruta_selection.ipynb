## Sélection des variables les plus informatives avec Boruta

Pour m'assurer d'utiliser les caractéristiques les plus robustes dans mon modèle, je lance Boruta.  
Cet algorithme va tester chaque variable du jeu de données contre de fausses colonnes (aléatoires) et ne garder que celles qui apportent une vraie plus-value à la séparation entre classes.  
C'est une étape clé après la visualisation et l'ACP pour justifier mes choix de variables à garder.


```python
# 1. Import et préparation des données
from sklearn.datasets import load_breast_cancer
import pandas as pd
import numpy as np

data = load_breast_cancer()
X = data.data                  # variables explicatives
y = data.target                # cible
feature_names = data.feature_names

# 2. Standardisation indispensable pour Boruta
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("Dimensions X_scaled :", X_scaled.shape)
print("Exemple y :", y[:10])

# 3. Sélection automatique de variables avec Boruta
from sklearn.ensemble import RandomForestClassifier
from boruta import BorutaPy

rf = RandomForestClassifier(n_estimators=100, random_state=42)
boruta_selector = BorutaPy(rf, n_estimators='auto', random_state=42)
boruta_selector.fit(X_scaled, y)

print("Variables importantes retenues par Boruta :")
print(np.array(feature_names)[boruta_selector.support_])
```

### Résultat et ouverture

Boruta me donne une liste objective de variables importantes ; je peux les comparer aux variables "visuellement intéressantes" (via ACP/heatmap).
